```python
import pandas as pd
import numpy as np
```

#Restaurant Chain Analysis

```python
restaurant_data = {
    'Branch': ['Downtown', 'Downtown', 'Suburb', 'Suburb', 'Airport', 'Airport',
               'Downtown', 'Suburb', 'Airport', 'Downtown', 'Suburb', 'Airport'],
    'Day': ['Mon', 'Tue', 'Mon', 'Tue', 'Mon', 'Tue',
            'Wed', 'Wed', 'Wed', 'Thu', 'Thu', 'Thu'],
    'Revenue': [5000, 5500, 3000, 3200, 4000, 4500,
                5200, 3100, 4200, 5300, 3300, 4400],
    'Customers': [120, 130, 80, 85, 95, 100,
                  125, 82, 98, 128, 88, 102],
    'Rating': [4.5, 4.6, 4.2, 4.3, 4.4, 4.5,
               4.7, 4.2, 4.6, 4.6, 4.4, 4.7]
}


restaurant_df = pd.DataFrame(restaurant_data)
print("Restaurants data frame:")
print(restaurant_df)
```

```python
restaurant_df['Branch_Avg_Revenue'] = restaurant_df.groupby('Branch')['Revenue'].transform('mean')
print("\nRestaurants data frame with branch average revenue:")
print(restaurant_df)
```

```python
restaurant_df['Revenue_vs_Avg'] = restaurant_df['Revenue'] - restaurant_df['Branch_Avg_Revenue']
print("\nRestaurants data frame with revenue vs branch average:")
print(restaurant_df)
```

```python
filter = restaurant_df.groupby('Branch')['Revenue'].transform('mean') > 4000
print("\nRestaurants data frame with branch average revenue greater than 4000:")
print(restaurant_df[filter])
```

```python
total_revenue = restaurant_df.groupby('Branch')['Revenue'].sum()
print("\nTotal revenue:", total_revenue)
```

```python
avg_customers = restaurant_df.groupby('Branch')['Customers'].mean()
print("\nAverage customers:", avg_customers)
```

```python
max_rating = restaurant_df.groupby('Branch')['Rating'].max()
print("\nMaximum rating:", max_rating)
```

```python
restaurant_df['Revenue_Rank'] = restaurant_df.groupby('Day')['Revenue'].rank(ascending=False)
print("\nRestaurants data frame with revenue rank:")
print(restaurant_df)
```

#Employee Performance Tracking

```python
employee_performance = {
    'Employee': ['Ali', 'Ali', 'Ali', 'Vali', 'Vali', 'Vali',
                 'Guli', 'Guli', 'Guli', 'Sami', 'Sami', 'Sami'],
    'Department': ['Sales', 'Sales', 'Sales', 'IT', 'IT', 'IT',
                   'Sales', 'Sales', 'Sales', 'IT', 'IT', 'IT'],
    'Month': ['Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar',
              'Jan', 'Feb', 'Mar', 'Jan', 'Feb', 'Mar'],
    'Tasks_Completed': [15, 18, 20, 12, 14, 16, 16, 17, 19, 11, 13, 15],
    'Hours_Worked': [160, 165, 170, 155, 160, 165, 162, 168, 172, 150, 155, 160],
    'Performance_Score': [85, 88, 92, 78, 82, 86, 87, 89, 91, 75, 79, 83]
}

employee_df = pd.DataFrame(employee_performance)
print("Employee performance data frame:")
print(employee_df)
```

```python
productivity = employee_df['Tasks_Completed'] * employee_df['Hours_Worked'] * 100
employee_df['Productivity'] = productivity
print("\nEmployee performance data frame with productivity:")
print(employee_df)
```

```python
each_employee_productivity = employee_df.groupby('Employee')['Productivity'].sum()
print("\nEach employee's productivity:")
print(each_employee_productivity)
```

```python
employee_df['Dept_Avg_Score'] = employee_df.groupby('Department')['Performance_Score'].transform('mean')
print("\nEmployee performance data frame with department average score:")
print(employee_df)
```

```python
employee_df['Status'] = np.where(employee_df['Performance_Score'] > employee_df['Dept_Avg_Score'], 'Above Average', 'Below Average')
print("\nEmployee performance data frame with status:")
print(employee_df)
```

```python
each_employee_three_month_productivity = employee_df.groupby(['Employee', 'Month'])['Productivity'].sum()
print("\nEach employee's productivity in the last three months:")
print(each_employee_three_month_productivity)
```

```python
employee_df['Performance_Improvement'] = employee_df['Performance_Score'] - employee_df['Dept_Avg_Score']
filtered_employees = employee_df[employee_df['Performance_Improvement'] > 5]
print("\nEmployees with Performance Improvement > 5:")
print(filtered_employees)
```

```python
employee_df['Department_Monthly_Rank'] = employee_df.groupby(['Department', 'Month'])['Performance_Score'].rank(ascending=False)
print("\nEmployee performance data frame with Department Monthly Rank:")
print(employee_df)
```

#Online Course Platform Analytics

```python
course_data = {
    'Course_ID': ['C1', 'C1', 'C2', 'C2', 'C3', 'C3', 'C4', 'C4'],
    'Category': ['Programming', 'Programming', 'Design', 'Design',
                 'Business', 'Business', 'Programming', 'Programming'],
    'Week': [1, 2, 1, 2, 1, 2, 1, 2],
    'Enrollments': [120, 150, 80, 95, 60, 70, 100, 130],
    'Completions': [100, 125, 70, 85, 55, 65, 85, 110],
    'Revenue': [6000, 7500, 4000, 4750, 3000, 3500, 5000, 6500]
}

course_df = pd.DataFrame(course_data)
print("Course data frame:")
print(course_df)
```

```python
course_df['Completion_Rate'] = (course_df['Completions'] / course_df['Enrollments']) * 100
print("\nCourse data frame with Completion Rate:")
print(course_df)
```

```python
course_df['Category_Avg_Revenue'] = course_df.groupby('Category')['Revenue'].transform('mean')
course_df['Category_Avg_Completion_Rate'] = course_df.groupby('Category')['Completion_Rate'].transform('mean')
print("\nCourse data frame with Category Average Revenue and Completion Rate:")
print(course_df)
```

```python
course_df['Revenue_Diff_%'] = ((course_df['Revenue'] - course_df['Category_Avg_Revenue']) / course_df['Category_Avg_Revenue']) * 100
print("\nCourse data frame with Revenue Difference %:")
print(course_df)
```

```python
category_summary = course_df.groupby('Category').apply(lambda x: pd.Series({
    'Total Enrollments': x['Enrollments'].sum(),
    'Total Revenue': x['Revenue'].sum(),
    'Best Completion Rate': x['Completion_Rate'].max()
}))
print("\nCategory Summary DataFrame:")
print(category_summary)
```

```python
filtered_categories = category_summary[category_summary['Best Completion Rate'] > 80]
print("\nCategories with Average Completion Rate > 80%:")
print(filtered_categories)
```

```python
allowed_categories = filtered_categories.index.tolist()
ranks_df = course_df[course_df['Category'].isin(allowed_categories)].copy()
ranks_df['Category_Revenue_Rank'] = ranks_df.groupby('Category')['Revenue'].rank(ascending=False)

print("\nCourses ranked by revenue within filtered categories:")
print(ranks_df.sort_values(by=['Category', 'Category_Revenue_Rank']))
```
